### Calculate the total sum of coins in Euros

```python
cent1 = int(input("Introduce the number of coins of 1 cent: ")) 
cent2 = int(input("Introduce the number of coins of 2 cents: ")) 
cent5 = int(input("Introduce the number of coins of 5 cents: ")) 
cent10 = int(input("Introduce the number of coins of 10 cents: ")) 
cent20 = int(input("Introduce the number of coins of 20 cents: ")) 
cent50 = int(input("Introduce the number of coins of 50 cents: ")) 

```

```python
total_value_in_euros = int(cent1)*0.01 + int(cent2)*0.02 + int(cent5)*0.05 + int(cent10)*0.10 + int(cent20)*0.20 + int(cent50)*0.50
print(total_value_in_euros , str("euros in total"))
```
