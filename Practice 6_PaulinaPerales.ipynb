## Program 1: Supress vowels

```python
the_string = str(input("Write a string: "))

#Lenght of the phrase
long=len(the_string)
vowels = ["a","e","i","o","u"]

output_sentence = ""

for letter in the_string:
    if letter not in vowels:
        output_sentence=output_sentence+letter
        
print("output sentence: " + output_sentence)
```

## Program 4: Create the second string by extracting

```python
#Ask the user to write a string
my_string = str(input("Introduce a line of text: "))
my_string_2 = ""

for word in my_string.split():
    if word[0].isupper():  
        my_string_2 += word[0] 

print ("Given string: ", my_string)
print("Created string: ", my_string_2) 
```
