# Section 1

At this point, we will create the **function CreateLists(patients, age, obesity, cholesterol, pat_num)**. This function is responsible for requesting data (Name, age, obesity, cholesterol) for each of the patients (there are num_pac number of patients) and storing them in the respective lists.

Once we have the function coded, we can proceed to **section 5 (main)**: Create the following empty lists: names, age, obesity, and cholesterol, and request the number of patients (pat_num) via keyboard input. The minimum number of patients will be 3 (**therefore it is necessary to VALIDATE that the number of patients entered is greater than or equal to 3. The request must be repeated until the condition is met**). Then call the function CreateLists(names, age, obesity, cholesterol, pat_num).
- `names` (string)
- `age` (int)
- `obesity` (string) values (YES/NO)
- `cholesterol` (int)

```python
def CreateLists(patients, age, obesity, cholesterol, pat_num):
    for i in range(pat_num):
        curr_patient = input(f"Give me the name for patient {i}: ")
        patients.append(curr_patient)
        
        curr_age = int(input(f"Give me the age for patient {i}: "))
        age.append(curr_age)
        
        curr_obesity = input(f"Tell me if patient {i} has obesity: ")
        obesity.append(curr_obesity)
        
        curr_cholesterol = input(f"Give me the cholesterol for patient {i}: ")
        cholesterol.append(curr_cholesterol)
    
    return
 
```

```python
patients = []
age = []
obesity = []
cholesterol = []

pat_num = int(input('Give me the number of patients: '))
while pat_num <3:
    pat_num = int(input('Give me the number of patients: '))
    
CreateLists(patients, age, obesity, cholesterol, pat_num)
```

```python
def CreateFile(file, sick_dict):
    file.write(f"name;age;obesity;cholesterol\n")
    for name, data in sick_dict.items():
        file.write(f"{name};{data[0]};{data[1]};{data[2]}\n")
    return
```

At this point, we will create the **function CalculateShow(file)**. This function reads the file, calculates the risk of heart attack for each patient, and displays on the screen the name of each patient and whether they are at risk of heart attack or not.

The risk indicator is "YES" if the patient meets the following criteria:
   * Is over 65 years old and has a very high level of cholesterol

Otherwise, "NO" will be displayed.

Cholesterol levels are described in the following table:

| Cholesterol Level | Category   |
|-------------------|------------|
| Less than 100     | Optimal    |
| 100 - 129         | Normal     |
| 130 - 159         | High       |
| 160 and above     | Very High  |

Additionally, this function must calculate the percentage of patients who are at risk. For this, we will keep count of the number of patients in the file in the variable `total_num_pac` and the count of those at risk in `num_pac_risk`. To find the percentage, we will calculate the percentage as `num_pac_risk` multiplied by 100 and divided by `total_num_pac`. This function returns this percentage value.

Once we have the function coded, we can proceed to **section 5 (main)**:
The `sick.csv` file must be opened in the main program and call the function CalculateShow(file). Afterwards, it must be closed.
Finally, a message indicating the percentage of patients at risk should be printed, as shown in the execution example.

```python
def CalculateShow(file):
    file.readline() 
    num_pac_risk = 0
    total_num_pac = 0
    for line in file: 
        line_split = line.strip().split(";") 
        

        total_num_pac = total_num_pac + 1
        print(line_split)
        
        if int(line_split[1]) > 65 and int(line_split[3]) > 160:
            print(f"{line_split[0]}: YES")
            num_pac_risk = num_pac_risk + 1
        else:
            print(f"{line_split[0]}: NO")
            
    perc_patients = num_pac_risk/total_num_pac
    
    return perc_patients
    
```

```python
file = open("patients.txt", "r", encoding="utf8")
perc_patients = CalculateShow(file)
file.close()
print(round(perc_patients,2))
```
