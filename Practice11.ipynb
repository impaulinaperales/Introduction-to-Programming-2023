# Program 1: Create a text file

```python
def Crear():
    num_books = int(input("Enter the number of books you want to add to the file: "))
    
    with open("books.txt", "w") as file:
        for i in range(num_books):
            author = input("Enter the name of the author: ")
            title = input("Enter the title of the book: ")
            file.write(f"{author};{title}\n")

def Mostrar():
    with open("books.txt", "r") as file:
        for line in file:
            author, title = line.strip().split(";")
            print(f"Author: {author}, Title: {title}")

def Buscar():
    author_to_search = input("Enter the name of the author to search: ")
    found_books = []
    
    with open("books.txt", "r") as file:
        for line in file:
            author, title = line.strip().split(";")
            if author == author_to_search:
                found_books.append(title)
    
    if found_books:
        print(f"Books by {author_to_search}: {', '.join(found_books)}")
    else:
        print(f"No books found for author {author_to_search}")

def Menu():
    while True:
        menu = """
    1) Display content of the file.
    2) Search for books of an author by name of the author. 
    3) Exit."""
        
        print(menu)
        choice = input("Enter your choice (1, 2, or 3): ")
        
        if choice == "1":
            Mostrar()
        elif choice == "2":
            Buscar()
        elif choice == "3":
            print("GoodBye!")
            break
        else:
            print("Invalid choice. Please enter 1, 2, or 3.")

if __name__ == "__main__":
    Crear()
    Menu()
```

# Program-2: create a text file of the city temperatures

```python
file = open("temps.txt","w+")

num_cities = int(input("Introduce the numer of cities: "))
for i in range(num_cities):
    city=input("Enter a city: ")
    temp= input("Enter the average temperature: ")
    file.write(city+"\t"+temp+"\n")
    
file.seek(0)
content = file.read()
print(content)

average_list = []
cities_list = []

info = content.split("\n")
info = info[:-1]

for line in info:
    citi_avg = line.split("\t")
    average_list.append(citi_avg[1])
    cities_list.append(citi_avg[0])
max_avg = max(average_list)
maximun_cities = []
k=0
for avg in average_list:
    if avg==max_avg:
        maximun_cities.append(cities_list[k])
    k=k+1
    
print(f"The city with the highest average temperature is {maximun_cities} with an average of {max_avg}")
    
```
