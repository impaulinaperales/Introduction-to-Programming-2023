## Program 2: Body mass index

```python
# Requests the weight in kg and the height in Cm.
```

```python
while True:
    weight = float(input("Write your weight in Kg: "))
    if weight > 0:
        break
    else:
        print("Error")
```

```python
while True:
    height = float(input("Write your height in Cm: "))
    if height > 0:
        break
    else:
        print("Error")
        
#The program asks you to give the height using cm but we have to convert it to metres
# in order to use the formula correctly

height = height/100
print(height)
```

```python
#Compute the BMI by using the  equation.

imc = weight/height**2
print(imc)
```

```python
#Display the category as defined by the BMI.

if imc<16:
    print("Severely underweight")
elif imc>16 and imc<18.5:
    print("Underweight")
elif imc>18.5 and imc<25:
    print("Normal")
elif imc>25 and imc<30:
    print("Overweight")
elif imc>30:
    print("Obese")
```
