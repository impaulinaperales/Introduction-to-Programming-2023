## Program 1: Customers of a bank

```python
# Ask for the number of customers
number_of_clients = int(input("Enter the number of clients: "))

# Validate number of clients
if number_of_clients <= 0:
  print("Number of customers must be a positive integer.")
else:
  valid_input = True

client_code = []
balance_list = []

    
for i in range(number_of_clients):
    client_code.append(int(input(f'Enter the code of customer {i+1}: ')))
    balance_list.append(float(input(f'Enter the balance of customer {i+1}: ')))
    print(client_code)
    print(balance_list)
    
# Display the menu
    
print("------ MENU------")
print("Option1) Deposit or withdraw money")
print("Option2) Search for customers with balances above a limit")
print("Option3) View information stored in code and balance lists")
print("Option4) Exit")
print("\n")

option = int(input('Enter an option from the menu: '))
valid_options = [1, 2, 3, 4]
while option not in valid_options:
    option = int(input('Enter an option from the menu: '))

    
if option == 1:
     account_code = int(input('Enter a client code: '))
     while account_code not in client_code:
         account_code = int(input('Enter a client code: '))
     print('1. Withdraw money')
     print('2. Deposit money')
     withdraw_or_deposit = float(input('Select an option: '))

     if withdraw_or_deposit == 1:
         withdraw_option = float(input('Enter amount to withdraw: '))
        
         while withdraw_option <= balance_list[client_code.index(account_code)]:
             balance_list[client_code.index(account_code)] -= withdraw_option
         while withdraw_option > balance_list[client_code.index(account_code)]:
             print('Please enter a smaller amount.')
             withdraw_option = float(input('Enter amount to withdraw: '))
            
     if withdraw_or_deposit == 2:
         deposit_option = float(input('Enter amount to deposit: '))
         balance_list[client_code.index(account_code)] += deposit_option
         print(f'New balance for customer {account_code}: {balance_list[client_code.index(account_code)]}')


elif option == 2:
     limit = float(input('Enter the balance limit: '))
     cl_rich = [] 

     for i in range(number_of_clients):
         if balance_list[i] > limit:
             cl_rich.append(client_code[i])

     if not cl_rich:
         print('No clients found with balances above the limit.')
     else:
         print(f'Clients with balances above the limit: {cl_rich}')
        

elif option == 3:
     print('Customer Codes:', client_code)
     print('Customer Balances:', balance_list)

elif option == 4:
     print('Exit the program.')
```

## Program 2: Student Grades

```python
# We create empty lists
students = []
grades = []

# Ask the name of the students
number_of_students = int(input("Write down the number of students: "))

# Ask for the grades and names of the students
for i in range(number_of_students):
    names = input("Write down the name of the student: ")
    grade = float(input("Write down the grade of the student {}: ".format(names)))
    students.append(names)
    grades.append(grade)
    
    print(students)
    print(grades)
    

while True:
    print("Menu:")
    print("a) Calculate the average grade of the subject.")
    print("b) Find the maximum grade and show who has obtained it.")
    print("c) Find the minimum grade and show who has obtained it.")
    print("d) Exit.")

    choice = input("Enter your choice (a, b, c, or d): ")

    if choice == "a":
        average = sum(grades) / len(grades)
        print("The average in the class is: {}".format(average))
    elif choice == "b":
        max_grade = max(grades)
        print(f'{students[grades.index(max_grade)]} has the maximun grade with {grades[grades.index(max_grade)]} points')
    elif choice == "c":
        min_grade = min(grades)
        print(f'{students[grades.index(min_grade)]} has the minimun grade with {grade[grades.index(min_grade)]} points')
    elif choice == "d":
        print("Exiting the program.")
        break
    else:
        print("Invalid choice. Please enter a valid option.")
        
```
